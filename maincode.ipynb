## Advanced Options

Beyond local editing two things are quite important: editing the notebooks in the markdown format and running Jupyter remotely.
The latter matters when we want to run the code on a faster server.
The former matters since Jupyter's native ipynb format stores a lot of auxiliary data that is
irrelevant to the content,
mostly related to how and where the code is run.
This is confusing for Git, making
reviewing contributions very difficult.
Fortunately there is an alternative---native editing in the markdown format.

### Markdown Files in Jupyter

If you wish to contribute to the content of this book, you need to modify the
source file (md file, not ipynb file) on GitHub.
Using the notedown plugin we
can modify notebooks in the md format directly in Jupyter.


First, install the notedown plugin, run the Jupyter Notebook, and load the plugin:

```
pip install d2l-notedown  # You may need to uninstall the original notedown.
jupyter notebook --NotebookApp.contents_manager_class='notedown.NotedownContentsManager'
```

You may also turn on the notedown plugin by default whenever you run the Jupyter Notebook.
First, generate a Jupyter Notebook configuration file (if it has already been generated, you can skip this step).

```
jupyter notebook --generate-config
```

Then, add the following line to the end of the Jupyter Notebook configuration file (for Linux or macOS, usually in the path `~/.jupyter/jupyter_notebook_config.py`):

```
c.NotebookApp.contents_manager_class = 'notedown.NotedownContentsManager'
```

After that, you only need to run the `jupyter notebook` command to turn on the notedown plugin by default.

### Running Jupyter Notebooks on a Remote Server

Sometimes, you may want to run Jupyter notebooks on a remote server and access it through a browser on your local computer. If Linux or macOS is installed on your local machine (Windows can also support this function through third-party software such as PuTTY), you can use port forwarding:

```
ssh myserver -L 8888:localhost:8888
```

The above string `myserver` is the address of the remote server.
Then we can use http://localhost:8888 to access the remote server `myserver` that runs Jupyter notebooks. We will detail on how to run Jupyter notebooks on AWS instances
later in this appendix.

### Timing

We can use the `ExecuteTime` plugin to time the execution of each code cell in Jupyter notebooks.
Use the following commands to install the plugin:

```
pip install jupyter_contrib_nbextensions
jupyter contrib nbextension install --user
jupyter nbextension enable execute_time/ExecuteTime
```

````python
# @title
class RobertaBiLSTMAttention(nn.Module):
    def __init__(self, hidden_dim=128, num_labels=2):
        super().__init__()
        self.roberta = RobertaModel.from_pretrained("roberta-base")

        # 🔥 Freeze ALL RoBERTa layers — Faster, Less RAM Usage
        for param in self.roberta.parameters():
            param.requires_grad = False

        self.lstm = nn.LSTM(
            input_size=self.roberta.config.hidden_size,
            hidden_size=hidden_dim,
            batch_first=True,
            bidirectional=True
        )
        self.attn = nn.Linear(hidden_dim * 2, 1)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(hidden_dim * 2, num_labels)

    def forward(self, input_ids, attention_mask):
        # 🔥 No gradients needed in RoBERTa
        with torch.no_grad():
            roberta_out = self.roberta(
                input_ids=input_ids,
                attention_mask=attention_mask
            ).last_hidden_state

        lstm_out, _ = self.lstm(roberta_out)
        weights = torch.softmax(self.attn(lstm_out), dim=1)
        context = torch.sum(weights * lstm_out, dim=1)
        output = self.fc(self.dropout(context))
        return output
````

````python
# Single-file runnable script / cell:
# Roberta + BiLSTM + Attention (RoBERTa frozen) with FP16 training, safe device handling,
# robust tokenization, checkpointing, evaluation, and Gradio serving.
# Paste into one notebook cell or save as a .py and run inside a proper environment.

import os
import platform
import warnings
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split
from torch.utils.data import DataLoader
from datasets import Dataset, DatasetDict
from transformers import RobertaTokenizer, RobertaModel, get_linear_schedule_with_warmup
from sklearn.metrics import accuracy_score
import gradio as gr

# ----------------------------
# Config
# ----------------------------
DATA_PATH = "/content/IMDB Dataset.csv"
MODEL_NAME = "roberta-base"
MAX_LEN = 128
BATCH_SIZE = 32
N_EPOCHS = 3
HIDDEN_DIM = 128
LR_LSTM = 3e-4                # LR for trainable parts (LSTM/FC)
CHECKPOINT_PATH = "best_model.pt"
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Python:", platform.python_version())
print("PyTorch:", torch.__version__)
print("Device:", DEVICE)

# optional: silence noisy future warnings from torch. (optional)
warnings.filterwarnings("ignore", category=FutureWarning, module="torch.cuda.amp")

# ----------------------------
# Load & preprocess CSV robustly
# ----------------------------
df = pd.read_csv(DATA_PATH, on_bad_lines='skip', engine='python')
# support common column names
if "review" in df.columns and "sentiment" in df.columns:
    df = df.rename(columns={"review": "text", "sentiment": "label"})
elif "text" in df.columns and "label" in df.columns:
    pass
else:
    raise ValueError("CSV must contain 'review'/'text' and 'sentiment'/'label' columns.")

# map labels if necessary
if df["label"].dtype == object:
    df["label"] = df["label"].map({"negative": 0, "positive": 1})

df = df.dropna(subset=["text", "label"]).reset_index(drop=True)
print("Loaded dataset shape:", df.shape)
print(df["label"].value_counts())

train_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df["label"])
train_ds = Dataset.from_pandas(train_df.reset_index(drop=True))
test_ds = Dataset.from_pandas(test_df.reset_index(drop=True))
dataset = DatasetDict({"train": train_ds, "test": test_ds})

# ----------------------------
# Tokenizer + robust tokenization
# ----------------------------
tokenizer = RobertaTokenizer.from_pretrained(MODEL_NAME)

def tokenize_batch(examples):
    # examples: dict of lists
    texts = examples.get("text") or examples.get("review")
    if texts is None:
        raise ValueError("No text/review column found in dataset batch.")
    return tokenizer(texts, padding="max_length", truncation=True, max_length=MAX_LEN)

# decide which columns actually exist and remove only those
cols_to_remove = [c for c in ("text", "review", "__index_level_0__") if c in dataset["train"].column_names]
encoded_dataset = dataset.map(tokenize_batch, batched=True, remove_columns=cols_to_remove)
# keep only input_ids, attention_mask, label if present
keep_cols = [c for c in ["input_ids", "attention_mask", "label"] if c in encoded_dataset["train"].column_names]
encoded_dataset.set_format(type="torch", columns=keep_cols)
print("Encoded dataset columns:", encoded_dataset["train"].column_names)

# ----------------------------
# Model definition (frozen RoBERTa)
# ----------------------------
class RobertaBiLSTMAttention(nn.Module):
    def __init__(self, model_name=MODEL_NAME, hidden_dim=HIDDEN_DIM, num_labels=2):
        super().__init__()
        # load roberta; ignore mismatched sizes for quietness
        self.roberta = RobertaModel.from_pretrained(model_name, ignore_mismatched_sizes=True)
        # freeze roberta parameters by default
        for p in self.roberta.parameters():
            p.requires_grad = False

        self.lstm = nn.LSTM(
            input_size=self.roberta.config.hidden_size,
            hidden_size=hidden_dim,
            num_layers=1,
            batch_first=True,
            bidirectional=True
        )
        self.attn = nn.Linear(hidden_dim * 2, 1)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(hidden_dim * 2, num_labels)

    def forward(self, input_ids, attention_mask):
        # make sure RoBERTa runs on same device as inputs (no grads required)
        with torch.no_grad():
            roberta_out = self.roberta(input_ids=input_ids, attention_mask=attention_mask).last_hidden_state
        lstm_out, _ = self.lstm(roberta_out)  # (batch, seq, 2*hidden)
        attn_w = torch.softmax(self.attn(lstm_out), dim=1)  # (batch, seq, 1)
        context = torch.sum(attn_w * lstm_out, dim=1)      # (batch, 2*hidden)
        out = self.fc(self.dropout(context))
        return out

# ----------------------------
# Instantiate model and ensure device placement
# ----------------------------
model = RobertaBiLSTMAttention().to(DEVICE)
# explicit ensure submodules moved
model.roberta = model.roberta.to(DEVICE)
print("Sample parameter device:", next(model.parameters()).device)
# sanity check: roberta embeddings device
print("roberta embedding device:", model.roberta.embeddings.word_embeddings.weight.device)

# ----------------------------
# DataLoaders
# ----------------------------
train_loader = DataLoader(encoded_dataset["train"], batch_size=BATCH_SIZE, shuffle=True, pin_memory=True, num_workers=2)
test_loader  = DataLoader(encoded_dataset["test"], batch_size=BATCH_SIZE, shuffle=False, pin_memory=True, num_workers=2)

# ----------------------------
# Optimizer, loss, scaler, scheduler
# ----------------------------
# Only train parameters that require grad (LSTM/attn/fc)
trainable_params = list(filter(lambda p: p.requires_grad, model.parameters()))
optimizer = torch.optim.AdamW(trainable_params, lr=LR_LSTM)
criterion = nn.CrossEntropyLoss()
scaler = torch.cuda.amp.GradScaler() if DEVICE.type == "cuda" else None

total_steps = N_EPOCHS * len(train_loader)
scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps=100, num_training_steps=total_steps)

# ----------------------------
# Evaluate function
# ----------------------------
def evaluate(model, loader, device=DEVICE):
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for batch in loader:
            # move batch to device
            input_ids = batch["input_ids"].to(device)
            attention_mask = batch["attention_mask"].to(device)
            labels = batch["label"].to(device).long()

            logits = model(input_ids, attention_mask)
            preds = torch.argmax(logits, dim=1)
            all_preds.extend(preds.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())
    return accuracy_score(all_labels, all_preds)

# ----------------------------
# Training loop (FP16 if available), checkpointing
# ----------------------------
best_acc = 0.0
for epoch in range(1, N_EPOCHS + 1):
    model.train()
    running_loss = 0.0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}/{N_EPOCHS}", leave=False)
    for step, batch in enumerate(pbar, start=1):
        # move batch to device
        input_ids = batch["input_ids"].to(DEVICE)
        attention_mask = batch["attention_mask"].to(DEVICE)
        labels = batch["label"].to(DEVICE).long()

        # quick device-assert for debugging (helps catch mismatches early)
        assert next(model.parameters()).device == input_ids.device, \
            f"Model device {next(model.parameters()).device} != input device {input_ids.device}"

        optimizer.zero_grad()
        if DEVICE.type == "cuda" and scaler is not None:
            with torch.cuda.amp.autocast():
                logits = model(input_ids, attention_mask)
                loss = criterion(logits, labels)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            logits = model(input_ids, attention_mask)
            loss = criterion(logits, labels)
            loss.backward()
            optimizer.step()

        # scheduler step after optimizer step
        scheduler.step()

        running_loss += loss.item()
        avg_loss = running_loss / step
        pbar.set_postfix(loss=f"{avg_loss:.4f}")

    # validation
    val_acc = evaluate(model, test_loader)
    print(f"Epoch {epoch} done. Val Acc: {val_acc:.4f}, Avg Loss: {avg_loss:.4f}")

    # checkpoint best
    if val_acc > best_acc:
        best_acc = val_acc
        torch.save({
            "model_state": model.state_dict(),
            "optimizer_state": optimizer.state_dict(),
            "scheduler_state": scheduler.state_dict(),
            "best_acc": best_acc,
            "epoch": epoch
        }, CHECKPOINT_PATH)
        print(f"Saved best checkpoint to {CHECKPOINT_PATH} (acc={best_acc:.4f})")

# ----------------------------
# Load best checkpoint (optional) and prepare predict + gradio
# ----------------------------
if os.path.exists(CHECKPOINT_PATH):
    ckpt = torch.load(CHECKPOINT_PATH, map_location=DEVICE)
    model.load_state_dict(ckpt["model_state"])
    print("Loaded best model from checkpoint; best_acc =", ckpt.get("best_acc"))

model = model.to(DEVICE)  # ensure on device

def predict_text(text):
    model.eval()
    tokens = tokenizer(text, return_tensors="pt", truncation=True, padding="max_length", max_length=MAX_LEN)
    input_ids = tokens["input_ids"].to(DEVICE)
    attention_mask = tokens["attention_mask"].to(DEVICE)
    with torch.no_grad():
        logits = model(input_ids, attention_mask)
        probs = torch.softmax(logits, dim=1)
        pred = torch.argmax(probs, dim=1).item()
        conf = probs[0][pred].item()
    return ("Positive" if pred == 1 else "Negative", float(conf))

# ----------------------------
# Gradio app (active)
# ----------------------------
def gradio_predict(text):
    label, conf = predict_text(text)
    return f"{label} ({conf*100:.1f}%)"

interface = gr.Interface(
    fn=gradio_predict,
    inputs=gr.Textbox(lines=4, placeholder="Enter movie review text here..."),
    outputs="text",
    title="RoBERTa + BiLSTM + Attention (Frozen RoBERTa)"
)

# Launch (share=True if you need public link)
interface.launch(share=False)
````
